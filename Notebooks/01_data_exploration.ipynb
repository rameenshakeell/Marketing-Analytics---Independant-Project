This notebook loads the marketing campaign dataset and explores its basic structure.
The dataset contains 2,240 customers of a retailer, with demographic info, spending across 6 product categories, and responses to 5 past marketing campaigns.

**Goal of this notebook:** Understand what we have before we clean or analyze anything.

```python
import pandas as pd

df = pd.read_csv('marketing_campaign.csv', sep='\t')

print("Shape of the data:", df.shape)
df.head()
```

## 1. Data Types & Structure

Before analyzing anything, we need to know how Python is interpreting each column.
Numbers should be numbers, dates should be dates, and text should be text.
This step will flag if something is wrong before our analysis. 

```python
df.info()
```

## 2. Missing Values

After looking at the data, we know Income has 24 missing values. Let's confirm and look at those rows directly to see if there's a pattern.

```python
# Count missing values per column
print("Missing values per column:")
print(df.isnull().sum())

print("\n" + "="*50 + "\n")

# Show the rows where Income is missing
print("Rows with missing Income:")
df[df['Income'].isnull()]
```

## 3. Handling Missing Income: Group-Based Imputation

After inspecting the 24 missing Income values (inspected across different dimensions: age, education, marital status, and spending behavior), the missing values appear randomly distributed with no demographic or behavioural patterns (Missing Completely at Random)

Rather than dropping the rows or using a single overall median, we'll impute by **education level** — assigning each missing customer the median income of customers with the same education.
This respects the structure of the data and gives more realistic estimates than a blanket fill.

```python
# Look at median income by education level before we impute
df.groupby('Education')['Income'].median()
```

### Imputing Missing Income

Now we fill missing Income values with the median income of customers sharing the same education level.


```python
# Fill missing Income with the median income of customers in the same education group
df['Income'] = df.groupby('Education')['Income'].transform(
    lambda x: x.fillna(x.median())
)

# Confirm no missing values remain
print("Missing Income values after imputation:", df['Income'].isnull().sum())
print("Total rows in dataset:", len(df))
```

## 4. Fixing the Date Column

`Dt_Customer` is currently stored as text instead of a proper date.
We'll convert it to datetime format so we can perform time-based analysis.

The dataset uses day-month-year format (DD-MM-YYYY), so we specify that explicitly to avoid ambiguity.

```python
# Convert Dt_Customer from text to a proper datetime
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')

# Confirm the conversion worked
print("Data type of Dt_Customer:", df['Dt_Customer'].dtype)
print("\nEarliest customer signup:", df['Dt_Customer'].min())
print("Latest customer signup:  ", df['Dt_Customer'].max())
print("\nDate range covers:", (df['Dt_Customer'].max() - df['Dt_Customer'].min()).days, "days")
```

## 5. Feature Engineering

Creating derived columns to make the data easier to analyze:

- **Age** calculated as of 2014 (when the data was collected)
- **Total_Spent** total dollars spent across all 6 product categories
- **Total_Purchases** total transactions across all channels
- **Children** total kids and teens in household
- **Has_Children** flag for households with any children
- **Customer_Tenure_Days** days since the customer joined (as of the latest signup date in the dataset)
- **Total_Campaigns_Accepted** total campaigns the customer responded to (out of 6)

These derived features will drive segmentation, profiling, and targeting analysis later.

```python
# 1. Age — calculated as of 2014 (data collection era)
df['Age'] = 2014 - df['Year_Birth']

# 2. Total dollars spent across all product categories
spending_cols = ['MntWines', 'MntFruits', 'MntMeatProducts',
                 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
df['Total_Spent'] = df[spending_cols].sum(axis=1)

# 3. Total purchases across all channels
purchase_cols = ['NumDealsPurchases', 'NumWebPurchases',
                 'NumCatalogPurchases', 'NumStorePurchases']
df['Total_Purchases'] = df[purchase_cols].sum(axis=1)

# 4. Children = kids + teens
df['Children'] = df['Kidhome'] + df['Teenhome']

# 5. Has_Children flag (1 if any children, 0 if none)
df['Has_Children'] = (df['Children'] > 0).astype(int)

# 6. Customer tenure in days, measured from the latest signup date in the dataset
reference_date = df['Dt_Customer'].max()
df['Customer_Tenure_Days'] = (reference_date - df['Dt_Customer']).dt.days

# 7. Total campaigns accepted (out of 6: Cmp1-5 + Response)
campaign_cols = ['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3',
                 'AcceptedCmp4', 'AcceptedCmp5', 'Response']
df['Total_Campaigns_Accepted'] = df[campaign_cols].sum(axis=1)

# Confirm: show the new columns for the first 5 customers
df[['Age', 'Total_Spent', 'Total_Purchases', 'Children',
    'Has_Children', 'Customer_Tenure_Days', 'Total_Campaigns_Accepted']].head()
```

```python
# Save the cleaned dataset to a new CSV in the project folder
df.to_csv('marketing_campaign_clean.csv', index=False)

# Confirm by listing the columns and final shape
print("Cleaned dataset saved as: marketing_campaign_clean.csv")
print("Final shape:", df.shape)
```
